```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Titanic dataset
dataset = sns.load_dataset('titanic')

# Display first few rows
print(dataset.head())

# Basic information
print(dataset.shape)
print(dataset.info())

# 1. Count plot: number of passengers by survival status
plt.figure(figsize=(6,4))
sns.countplot(x='survived', data=dataset)
plt.title('Count of Survived vs Not Survived')
plt.show()

# 2. Count plot: survival by gender
plt.figure(figsize=(6,4))
sns.countplot(x='sex', hue='survived', data=dataset)
plt.title('Survival Count by Gender')
plt.show()

# 3. Box plot: age distribution by gender
plt.figure(figsize=(6,4))
sns.boxplot(x='sex', y='age', data=dataset)
plt.title('Age Distribution by Gender')
plt.show()

# 4. Violin plot: age distribution by gender
plt.figure(figsize=(6,4))
sns.violinplot(x='sex', y='age', data=dataset)
plt.title('Age Distribution by Gender using Violin Plot')
plt.show()

# 5. Joint plot: age vs fare
sns.jointplot(x='age', y='fare', data=dataset, kind='scatter')
plt.show()

# 6. Correlation heatmap
corr = dataset.select_dtypes(include='number').corr()
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()

###########################################################################################
# Histogram of fare
plt.figure(figsize=(8,5))
sns.histplot(dataset['fare'], bins=10, kde=False)
plt.title('Distribution of Fare')
plt.xlabel('Fare')
plt.ylabel('Number of Passengers')
plt.show()
```
