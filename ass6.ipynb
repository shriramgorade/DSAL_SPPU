```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score
)

# Load iris dataset from CSV
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/iris.csv"
columns = ['sepal-length', 'sepal-width', 'petal-length', 'petal-width', 'class']
df = pd.read_csv(url, names=columns)

print("First 5 rows:")
print(df.head())

print("\nShape of dataset:")
print(df.shape)

print("\nInfo:")
print(df.info())

print("\nMissing values:")
print(df.isnull().sum())

# Split features and target
X = df.iloc[:, 0:4].values
y = df['class'].values

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

# Train Naive Bayes model
model = GaussianNB()
model.fit(X_train, y_train)

# Predict test set
y_pred = model.predict(X_test)

# Confusion matrix
cm = confusion_matrix(y_test, y_pred)
print("\nConfusion Matrix:")
print(cm)

# Overall evaluation
accuracy = accuracy_score(y_test, y_pred)
error_rate = 1 - accuracy
precision_macro = precision_score(y_test, y_pred, average='macro')
recall_macro = recall_score(y_test, y_pred, average='macro')

print("\nAccuracy:", accuracy)
print("Error Rate:", error_rate)
print("Precision (macro):", precision_macro)
print("Recall (macro):", recall_macro)

# Per-class TP, FP, TN, FN
classes = np.unique(y)
print("\nPer-class TP, FP, TN, FN:")

for i, cls in enumerate(classes):
    TP = cm[i, i]
    FP = cm[:, i].sum() - TP
    FN = cm[i, :].sum() - TP
    TN = cm.sum() - (TP + FP + FN)

    print(f"\nClass: {cls}")
    print("TP =", TP)
    print("FP =", FP)
    print("TN =", TN)
    print("FN =", FN)

    class_accuracy = (TP + TN) / cm.sum()
    class_precision = TP / (TP + FP) if (TP + FP) != 0 else 0
    class_recall = TP / (TP + FN) if (TP + FN) != 0 else 0
    class_error_rate = 1 - class_accuracy

    print("Accuracy =", class_accuracy)
    print("Error Rate =", class_error_rate)
    print("Precision =", class_precision)
    print("Recall =", class_recall)
```
