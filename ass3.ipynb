```python
import pandas as pd
import numpy as np
```

```python
# ---------------------------------------
# PART 1: Age group and income statistics
# ---------------------------------------
```

```python
df_salary = pd.DataFrame({
    'Age_Group' : ['18-25', '18-25', '26-35', '26-35', '36-45', '36-45', '46-55', '46-55'],
    'Income' : [25000, 30000, 40000, 45000, 50000, 55000, 60000, 65000]
})

print("Age group and income dataset:")
print(df_salary)
```

```python
# Summary statistics of income grouped by age group

grouped_income = df_salary.groupby('Age_Group')['Income'].agg(['mean', 'median', 'min', 'max', 'std'])
print("\nSummary statistics of Income grouped by Age_Group:")
print(grouped_income)
```

```python
# ---------------------------------------
# PART 2: Convert categorical responses to numeric list
# ---------------------------------------

# Example categorical responses
responses = ['Yes', 'No', 'Yes', 'Yes', 'No', 'No', 'Yes']

# Map categorical values to numeric values
response_numeric = {'Yes': 1, 'No': 0}
numeric_list = [response_numeric[val] for val in responses]

print("\nCategorical responses:")
print(responses)

print("\nNumeric list for responses:")
print(numeric_list)
```

```python
# ---------------------------------------
# PART 3: Iris dataset statistics by species
# ---------------------------------------

url = 'https://raw.githubusercontent.com/jbrownlee/Datasets/master/iris.csv'
col_name = ['sepal-length', 'sepal-width', 'petal-length', 'petal-width', 'class']

iris = pd.read_csv(url, names=col_name)

print("\nFirst 5 rows of Iris dataset:")
print(iris.head())

print("\nShape of Iris dataset:")
print(iris.shape)

print("\nBasic information about Iris dataset:")
print(iris.info())

print("\nOverall descriptive statistics of Iris dataset:")
print(iris.describe())

# Statistics grouped by species
iris_grouped_stats = iris.groupby('class').describe()

print("\nDescriptive statistics of Iris species:")
print(iris_grouped_stats)
```
